```python
import random

machines = ['A', 'B', 'C']
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0, 'B': 0, 'C': 0}

epsilon = 0.2

for i in range(3):
    r = random.random()

    if r < epsilon:
        m = random.choice(machines)
        print("Exploration")
    else:
        m = max(Q, key=Q.get)
        print("Exploitation")

    print("Play with machine", m)
    reward = float(input("Enter reward: "))

    N[m] += 1
    R[m].append(reward)

    Q[m] = Q[m] + (reward - Q[m]) / N[m]

print("\nN Table:", N)
print("R Table:", R)
print("Q Table:", Q)
```
